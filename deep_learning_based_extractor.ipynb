```python
# train_unet_resnet50_gray_smp.py
import os
import shutil
import random
from glob import glob
from typing import List, Tuple

import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm
from PIL import Image
import pathlib

# =========================
# Local caches to avoid permission issues (e.g., Kaggle / Colab)
# =========================
DL_DIR = pathlib.Path("./downloads")
for p in ["xdg", "torch", "huggingface", "matplotlib", "torchinductor_user"]:
    (DL_DIR / p).mkdir(parents=True, exist_ok=True)
os.environ.setdefault("USER", "user")
os.environ.setdefault("XDG_CACHE_HOME", str(DL_DIR / "xdg"))
os.environ.setdefault("TORCH_HOME",     str(DL_DIR / "torch"))
os.environ.setdefault("HF_HOME",        str(DL_DIR / "huggingface"))
os.environ.setdefault("TORCHINDUCTOR_CACHE_DIR", str(DL_DIR / "torchinductor_user"))

# =========================
# segmentation_models_pytorch import with soft auto-install
# =========================
def _import_smp():
    try:
        import segmentation_models_pytorch as smp  # type: ignore
        return smp
    except Exception:
        try:
            import sys, subprocess
            print("Installing 'segmentation-models-pytorch' and 'timm' ...")
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q",
                                   "segmentation-models-pytorch==0.3.3", "timm>=0.9.2"])
            import segmentation_models_pytorch as smp  # type: ignore
            return smp
        except Exception as e:
            raise ImportError(
                "Could not import/install 'segmentation-models-pytorch'. "
                "If you're offline, install it beforehand:\n"
                "  pip install segmentation-models-pytorch timm\n"
            ) from e

smp = _import_smp()

# =========================
# Configuration
# =========================
IMAGE_DIR = "/kaggle/input/camus-extracted/images/training"
MASK_DIR  = "/kaggle/input/camus-fanshape-masks/camus_fanshape_masks"

# Writable output
PRED_DIR  = "./predicted_training_masks"

# If offline, set to False to avoid pretrained encoder downloads.
PRETRAINED_ENCODER = True

IMG_SIZE   = 512
BATCH_SIZE = 8
EPOCHS = 200
FREEZE_ENCODER_EPOCHS = 3
EARLY_STOP_PATIENCE = 5
LR = 1e-4
WEIGHT_DECAY = 1e-4
SEED = 42
NUM_WORKERS = 2
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# =========================
# Reproducibility
# =========================
def set_seed(s: int = 42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)
    torch.cuda.manual_seed_all(s)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(SEED)

# =========================
# Helpers
# =========================
IMG_EXTS = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff", ".gif"}

def basename_noext(path: str):
    return os.path.splitext(os.path.basename(path))[0]

def find_with_same_basename(folder: str, base: str):
    """Return path to file in 'folder' whose basename (no ext) == base, else None."""
    for ext in IMG_EXTS:
        candidate = os.path.join(folder, base + ext)
        if os.path.isfile(candidate):
            return candidate
    for p in glob(os.path.join(folder, "*")):
        if basename_noext(p) == base and os.path.isfile(p):
            return p
    return None

def list_images(folder: str):
    files = []
    for ext in IMG_EXTS:
        files.extend(glob(os.path.join(folder, f"*{ext}")))
    files = sorted(list(set(files)))
    return files

def ensure_dir_clean(path: str):
    if os.path.isdir(path):
        shutil.rmtree(path)
    os.makedirs(path, exist_ok=True)

def ensure_dir(path: str):
    os.makedirs(path, exist_ok=True)

def load_mask_with_alpha(path: str, img_size: int) -> np.ndarray:
    """
    Load a mask and return a strictly binary array (H, W) in {0,1}.

    Supports:
      - White foreground on transparent background (uses alpha).
      - White foreground on black background.
      - Black foreground on white background (auto-inverts).
    """
    img = Image.open(path)

    # ---------- 1) Try to use alpha channel if it really encodes transparency ----------
    has_alpha = (img.mode in ("RGBA", "LA")) or ("transparency" in img.info)
    if has_alpha:
        rgba = img.convert("RGBA")
        r, g, b, a = rgba.split()

        if a.size != (img_size, img_size):
            a = a.resize((img_size, img_size), Image.NEAREST)

        a_np = np.array(a, dtype=np.uint8)

        # If alpha has some transparent and some opaque pixels → use it
        if (a_np.max() > 0) and (a_np.min() == 0) and (a_np.mean() < 255):
            m = (a_np > 0).astype(np.uint8)
            return m

        # Alpha is useless (e.g. fully opaque) → fall back to grayscale
        img = rgba.convert("L")
    else:
        # No alpha channel → use luminance directly
        img = img.convert("L")

    # ---------- 2) Grayscale thresholding for black/white masks ----------
    if img.size != (img_size, img_size):
        img = img.resize((img_size, img_size), Image.NEAREST)

    g = np.array(img, dtype=np.uint8)

    # Basic binarization: values above mid-gray considered foreground
    m = (g > 127).astype(np.uint8)

    # ---------- 3) Auto-invert when background is white and FG is black ----------
    # If mask is mostly 1s, treat 1 as background and invert
    if m.mean() > 0.5:
        m = 1 - m

    return m

# =========================
# Dataset
# =========================
class GraySegDataset(Dataset):
    """
    Labeled data (image+mask). Returns:
      img: (1,H,W) float32 in [0,1]
      msk: (1,H,W) float32 in {0,1}
    """
    def __init__(self, items: List[Tuple[str, str]], img_size: int = 512, augment: bool = True):
        self.items = items
        self.img_size = img_size
        self.augment = augment

    def __len__(self):
        return len(self.items)

    def _load_gray(self, path: str):
        img = Image.open(path).convert("L")
        if img.size != (self.img_size, self.img_size):
            img = img.resize((self.img_size, self.img_size), Image.BILINEAR)
        return np.array(img, dtype=np.uint8)

    def _load_mask(self, path: str):
        return load_mask_with_alpha(path, self.img_size)

    def _augment_pair(self, img: np.ndarray, msk: np.ndarray):
        # Horizontal flip
        if random.random() < 0.5:
            img = np.flip(img, axis=1).copy()
            msk = np.flip(msk, axis=1).copy()
        # Small rotation
        if random.random() < 0.7:
            angle = random.uniform(-10, 10)
            imP = Image.fromarray(img)
            mkP = Image.fromarray((msk * 255).astype(np.uint8))
            imP = imP.rotate(angle, resample=Image.BILINEAR, expand=False, fillcolor=0)
            mkP = mkP.rotate(angle, resample=Image.NEAREST,  expand=False, fillcolor=0)
            img = np.array(imP, dtype=np.uint8).copy()
            msk = (np.array(mkP, dtype=np.uint8) > 0).astype(np.uint8).copy()
        return img, msk

    def __getitem__(self, idx):
        img_path, mask_path = self.items[idx]
        img = self._load_gray(img_path)
        msk = self._load_mask(mask_path)
        if self.augment:
            img, msk = self._augment_pair(img, msk)

        img = np.ascontiguousarray(img)
        msk = np.ascontiguousarray(msk)

        img = torch.from_numpy(img).float().unsqueeze(0) / 255.0
        msk = torch.from_numpy(msk).float().unsqueeze(0)
        return img, msk, os.path.basename(img_path)

class GrayUnlabeledDataset(Dataset):
    """For inference on images without masks."""
    def __init__(self, image_paths: List[str], img_size: int = 512):
        self.image_paths = image_paths
        self.img_size = img_size

    def __len__(self):
        return len(self.image_paths)

    def _load_gray(self, path: str):
        img = Image.open(path).convert("L")
        if img.size != (self.img_size, self.img_size):
            img = img.resize((self.img_size, self.img_size), Image.BILINEAR)
        return np.array(img, dtype=np.uint8)

    def __getitem__(self, idx):
        p = self.image_paths[idx]
        arr = self._load_gray(p)
        arr = np.ascontiguousarray(arr)
        ten = torch.from_numpy(arr).float().unsqueeze(0) / 255.0
        return ten, os.path.basename(p), arr

# =========================
# Model, Loss, Metrics
# =========================
def build_model():
    try:
        model = smp.Unet(
            encoder_name="resnet50",
            encoder_weights="imagenet" if PRETRAINED_ENCODER else None,
            in_channels=1,
            classes=1,
        )
        return model
    except Exception as e:
        print(f"[build_model] Falling back to encoder_weights=None due to: {e}")
        model = smp.Unet(
            encoder_name="resnet50",
            encoder_weights=None,
            in_channels=1,
            classes=1,
        )
        return model

class DiceLoss(nn.Module):
    def __init__(self, smooth: float = 1e-6):
        super().__init__()
        self.smooth = smooth
    def forward(self, logits, targets):
        probs = torch.sigmoid(logits)
        num = 2 * torch.sum(probs * targets) + self.smooth
        den = torch.sum(probs) + torch.sum(targets) + self.smooth
        return 1 - (num / den)

def bce_dice_loss(logits, targets, bce_weight=0.5):
    bce = nn.functional.binary_cross_entropy_with_logits(logits, targets)
    dice = DiceLoss()(logits, targets)
    return bce_weight * bce + (1 - bce_weight) * dice

@torch.no_grad()
def dice_coef(logits, targets, thr=0.5):
    probs = torch.sigmoid(logits)
    preds = (probs > thr).float()
    num = 2 * torch.sum(preds * targets)
    den = torch.sum(preds) + torch.sum(targets) + 1e-6
    return (num / den).item()

# =========================
# Training Utils
# =========================
class EarlyStopping:
    def __init__(self, patience=5, mode="min", min_delta=0.0):
        self.patience = patience
        self.mode = mode
        self.min_delta = min_delta
        self.best = None
        self.num_bad = 0
        self.should_stop = False

    def step(self, value):
        if self.best is None:
            self.best = value
            return False
        improve = (value < self.best - self.min_delta) if self.mode == "min" else (value > self.best + self.min_delta)
        if improve:
            self.best = value
            self.num_bad = 0
        else:
            self.num_bad += 1
            if self.num_bad >= self.patience:
                self.should_stop = True
        return self.should_stop

def train_one_epoch(model, loader, optimizer, device):
    model.train()
    running_loss = 0.0
    running_dice = 0.0
    for imgs, masks, _ in tqdm(loader, desc="Train", leave=False):
        imgs = imgs.to(device, non_blocking=True)
        masks = masks.to(device, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        logits = model(imgs)
        loss = bce_dice_loss(logits, masks, bce_weight=0.5)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * imgs.size(0)
        running_dice += dice_coef(logits.detach(), masks)
    n = len(loader.dataset)
    return running_loss / n, running_dice / max(1, len(loader))

@torch.no_grad()
def validate_one_epoch(model, loader, device):
    model.eval()
    running_loss = 0.0
    running_dice = 0.0
    for imgs, masks, _ in tqdm(loader, desc="Valid", leave=False):
        imgs = imgs.to(device, non_blocking=True)
        masks = masks.to(device, non_blocking=True)
        logits = model(imgs)
        loss = bce_dice_loss(logits, masks, bce_weight=0.5)
        running_loss += loss.item() * imgs.size(0)
        running_dice += dice_coef(logits, masks)
    n = len(loader.dataset)
    return running_loss / n, running_dice / max(1, len(loader))

# =========================
# Discovery: labeled vs unlabeled
# =========================
def prepare_file_lists(img_dir: str, msk_dir: str):
    all_imgs = list_images(img_dir)
    labeled: List[Tuple[str, str]] = []
    unlabeled: List[str] = []
    for ip in all_imgs:
        base = basename_noext(ip)
        mp = find_with_same_basename(msk_dir, base)
        if mp is not None:
            labeled.append((ip, mp))
        else:
            unlabeled.append(ip)
    return labeled, unlabeled

# =========================
# Visualization
# =========================
@torch.no_grad()
def visualize_triptychs(model, device, items: List[Tuple[str,str]], max_samples=20, cols=4):
    model.eval()
    if len(items) == 0:
        print("No labeled items to visualize.")
        return
    sample_items = items[:max_samples]
    rows = len(sample_items)
    cols = min(cols, 4)
    fig, axes = plt.subplots(rows, cols, figsize=(cols*3.4, rows*3.2))
    if rows == 1:
        axes = np.expand_dims(axes, 0)

    for r, (ip, mp) in enumerate(sample_items):
        img = Image.open(ip).convert("L").resize((IMG_SIZE, IMG_SIZE), Image.BILINEAR)
        img_np = np.array(img, dtype=np.uint8)
        gt = load_mask_with_alpha(mp, IMG_SIZE)

        ten = torch.from_numpy(np.ascontiguousarray(img_np)).float().unsqueeze(0).unsqueeze(0).to(device) / 255.0
        logits = model(ten)
        prob = torch.sigmoid(logits)[0,0].cpu().numpy()
        pred = (prob > 0.5).astype(np.uint8)

        overlay = np.stack([img_np]*3, axis=-1)
        g = overlay[...,1].astype(np.float32)
        g[pred==1] = 255
        overlay[...,1] = g.clip(0,255).astype(np.uint8)

        ax_row = axes[r]
        ax0, ax1, ax2, ax3 = ax_row[0], ax_row[1], ax_row[2], ax_row[3]
        ax0.imshow(img_np, cmap="gray"); ax0.set_title("Image"); ax0.axis("off")
        ax1.imshow(gt, cmap="gray");     ax1.set_title("GT Mask"); ax1.axis("off")
        ax2.imshow(pred, cmap="gray", vmin=0, vmax=1);   ax2.set_title("Pred Mask (0/1)"); ax2.axis("off")
        ax3.imshow(overlay);             ax3.set_title("Overlay"); ax3.axis("off")

    plt.tight_layout()
    plt.show()

# =========================
# Inference helpers
# =========================
@torch.no_grad()
def predict_mask_from_imagepath(model, device, img_path: str, img_size: int = IMG_SIZE, thr: float = 0.5) -> np.ndarray:
    """Load grayscale image, resize, predict strictly binary mask (H,W) in {0,1}."""
    img = Image.open(img_path).convert("L")
    if img.size != (img_size, img_size):
        img = img.resize((img_size, img_size), Image.BILINEAR)
    arr = np.ascontiguousarray(np.array(img, dtype=np.uint8))
    ten = torch.from_numpy(arr).float().unsqueeze(0).unsqueeze(0).to(device) / 255.0
    logits = model(ten)
    prob = torch.sigmoid(logits)[0,0].detach().cpu().numpy()
    pred = (prob > thr).astype(np.uint8)  # strictly 0/1
    return pred

def save_pred_mask(mask_bin: np.ndarray, out_path: str):
    """Save strictly binary mask (values 0/1) to disk."""
    mask_bin = (mask_bin > 0).astype(np.uint8)
    Image.fromarray(mask_bin).save(out_path)

# =========================
# Main
# =========================
def main():
    print("Preparing data...")
    labeled, unlabeled = prepare_file_lists(IMAGE_DIR, MASK_DIR)
    print(f"Found {len(labeled)} labeled images (with masks) and {len(unlabeled)} unlabeled images.")

    if len(labeled) == 0:
        raise RuntimeError("No labeled pairs found. Ensure mask basenames match image basenames.")

    # Split
    set_seed(SEED)
    random.shuffle(labeled)
    val_count = max(1, int(0.2 * len(labeled)))
    val_items = labeled[:val_count]
    train_items = labeled[val_count:]
    print(f"Train: {len(train_items)} | Val: {len(val_items)}")

    train_ds = GraySegDataset(train_items, img_size=IMG_SIZE, augment=True)
    val_ds   = GraySegDataset(val_items,   img_size=IMG_SIZE, augment=False)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                              num_workers=NUM_WORKERS, pin_memory=True, drop_last=False)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                              num_workers=NUM_WORKERS, pin_memory=True, drop_last=False)

    print("Building model...")
    model = build_model().to(DEVICE)

    optimizer = optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", patience=2, factor=0.5, verbose=True)
    early = EarlyStopping(patience=EARLY_STOP_PATIENCE, mode="min")

    # Freeze encoder for warmup
    def set_encoder_trainable(is_trainable: bool):
        enc = getattr(model, "encoder", None)
        if enc is not None:
            for p in enc.parameters():
                p.requires_grad = is_trainable

    set_encoder_trainable(False)
    best_val = float("inf")
    best_path = "unet_resnet50_best.pth"

    print("Starting training...")
    for epoch in range(1, EPOCHS+1):
        print(f"\nEpoch {epoch}/{EPOCHS}")

        if epoch == FREEZE_ENCODER_EPOCHS + 1:
            print("Unfreezing encoder...")
            set_encoder_trainable(True)

        train_loss, train_dice = train_one_epoch(model, train_loader, optimizer, DEVICE)
        val_loss, val_dice     = validate_one_epoch(model, val_loader, DEVICE)
        scheduler.step(val_loss)

        print(f"Train Loss: {train_loss:.4f} | Train Dice: {train_dice:.4f}")
        print(f"Val   Loss: {val_loss:.4f} | Val   Dice: {val_dice:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            torch.save(model.state_dict(), best_path)
            print(f"  -> Saved new best to {best_path}")

        if early.step(val_loss):
            print(f"Early stopping triggered (patience={EARLY_STOP_PATIENCE}).")
            break

    # Save last
    last_path = "unet_resnet50_last.pth"
    torch.save(model.state_dict(), last_path)
    print(f"Saved last weights to {last_path}")

    # Load best for inference
    model.load_state_dict(torch.load(best_path, map_location=DEVICE))
    model.eval()

    # Prepare output folders
    print("\nPreparing output folders...")
    ensure_dir_clean(PRED_DIR)
    out_unlabeled = os.path.join(PRED_DIR, "unlabeled")
    out_train     = os.path.join(PRED_DIR, "train")
    out_val       = os.path.join(PRED_DIR, "val")
    ensure_dir(out_unlabeled)
    ensure_dir(out_train)
    ensure_dir(out_val)

    # Predict unlabeled -> ./predicted_training_masks/unlabeled/ (0/1 images)
    if len(unlabeled) > 0:
        print("\nPredicting unlabeled images...")
        unlabeled_ds = GrayUnlabeledDataset(unlabeled, img_size=IMG_SIZE)
        unlabeled_loader = DataLoader(unlabeled_ds, batch_size=1, shuffle=False,
                                      num_workers=NUM_WORKERS, pin_memory=True)

        with torch.no_grad():
            for (img_t, fname, _raw) in tqdm(unlabeled_loader, desc="Infer(unlabeled)", leave=False):
                img_t = img_t.to(DEVICE)
                logits = model(img_t)
                prob = torch.sigmoid(logits)[0,0].cpu().numpy()
                pred = (prob > 0.5).astype(np.uint8)  # strictly 0/1
                out_path = os.path.join(out_unlabeled, os.path.splitext(fname[0])[0] + ".png")
                save_pred_mask(pred, out_path)

        print(f"Saved {len(unlabeled)} predicted masks to: {out_unlabeled}")
    else:
        print("No unlabeled images found. Skipping unlabeled inference.")

    # Predict TRAIN images -> ./predicted_training_masks/train/ (0/1 images)
    print("\nPredicting TRAIN images (labeled set used for training)...")
    n_train_saved = 0
    with torch.no_grad():
        for (ip, _mp) in tqdm(train_items, desc="Infer(train)", leave=False):
            pred = predict_mask_from_imagepath(model, DEVICE, ip, img_size=IMG_SIZE, thr=0.5)
            out_path = os.path.join(out_train, basename_noext(ip) + ".png")
            save_pred_mask(pred, out_path)
            n_train_saved += 1
    print(f"Saved {n_train_saved} train predicted masks to {out_train}")

    # Predict VAL images -> ./predicted_training_masks/val/ (0/1 images)
    print("\nPredicting VAL images (held-out labeled set)...")
    n_val_saved = 0
    with torch.no_grad():
        for (ip, _mp) in tqdm(val_items, desc="Infer(val)", leave=False):
            pred = predict_mask_from_imagepath(model, DEVICE, ip, img_size=IMG_SIZE, thr=0.5)
            out_path = os.path.join(out_val, basename_noext(ip) + ".png")
            save_pred_mask(pred, out_path)
            n_val_saved += 1
    print(f"Saved {n_val_saved} val predicted masks to {out_val}")

    # Visual check
    print("\nDisplaying up to 20 labeled samples with GT and predictions...")
    labeled_sorted = sorted(labeled, key=lambda x: os.path.basename(x[0]))
    visualize_triptychs(model, DEVICE, labeled_sorted, max_samples=20, cols=4)

if __name__ == "__main__":
    main()
```
